# Bài tập 1: So sánh %timeit giữa List Comprehension và NumPy

So sánh tốc độ tạo 100.000 số ngẫu nhiên giữa Python thuần (`list comprehension`) và thư viện `numpy`.

```python
import random
import numpy as np
```

### 1.1 Đo tốc độ bằng List Comprehension (Python thuần)

```python
%timeit [random.random() for _ in range(100_000)]
```

### 1.2 Đo tốc độ bằng NumPy (`np.random.rand`)

```python
%timeit np.random.rand(100_000)
```

# Bài tập 2: Đọc CSV, hiển thị DataFrame và vẽ biểu đồ

Tạo file CSV mẫu, đọc vào DataFrame bằng `pandas` và trực quan hóa bằng `matplotlib`.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

# Tạo file CSV mẫu để notebook tự chạy được từ đầu đến cuối
csv_content = """Thang,DoanhThu,LoiNhuan
Thang 1,120,45
Thang 2,150,55
Thang 3,170,68
Thang 4,160,60
Thang 5,200,85
Thang 6,240,110"""

with open('sales_data.csv', 'w', encoding='utf-8') as f:
    f.write(csv_content.strip())

print('Da tao file sales_data.csv thanh cong!')
```

### 2.1 Đọc file CSV và hiển thị DataFrame

```python
df = pd.read_csv('sales_data.csv')
df
```

### 2.2 Vẽ biểu đồ đường trực quan hóa Doanh thu và Lợi nhuận

```python
plt.figure(figsize=(9, 5))
plt.plot(df['Thang'], df['DoanhThu'], marker='o', color='royalblue', label='Doanh thu (trieu VND)', linewidth=2)
plt.plot(df['Thang'], df['LoiNhuan'], marker='s', color='forestgreen', label='Loi nhuan (trieu VND)', linewidth=2)

plt.title('Bao cao Doanh thu va Loi nhuan 6 thang dau nam', fontsize=14, fontweight='bold')
plt.xlabel('Thoi gian', fontsize=11)
plt.ylabel('Gia tri (trieu VND)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
